```python
# I used AI to build a labeling tool so I could manually tag each tie's color.
# The labels are saved in Data/tie_labels.csv.

tie_color = pd.read_csv("Data/tie_labels.csv")
tie_color.head()
```

```python
# Count the occurrences of each tie color in the dataset

tie_color["tie_color"].value_counts()
```

```python
# Plot the top 10 most common tie colors in a horizontal bar chart

tie_color["tie_color"].value_counts().head(10).sort_values().plot.barh()
```

```python
# Calculate the proportion of red ties in the dataset. It's 36%!

(tie_color["tie_color"] == "red").mean()
```
